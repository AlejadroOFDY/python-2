### Normalización del dataset

5. Aplica las operaciones necesarias para normalizar los valores del dataset, para que se encuentren en el intervalo [0, 1].

```python
X_train = X_train / 255.0
X_test = X_test / 255.0
```

### Configuración de la red neuronal

6. Crea un modelo `Sequential` con Keras.
  - a) Configura la capa de entrada según el tamaño de las imágenes encontrado en el punto `2.`.
  - b) Configura una capa para aplanar las imágenes.
  - c) Añade **al menos dos capas densas**, con las dimensiones y funciones de activación que prefieras.
  - d) Si el problema tratado es la clasificación multiclase, ¿qué función de activación **es necesaria** en la capa de salida? Aplícala.  

```python
import keras

# Configura tu red neuronal
model = keras.Sequential([
    keras.layers.Input(shape=(28,28)),
    keras.layers.Flatten(),
    keras.layers.Dense(128, activation="relu"),
    keras.layers.Dense(64, activation="relu"),
    keras.layers.Dense(10, activation="softmax")
])
```

### Entrenamiento



7. Compila tu red neuronal, especificando:
   - a) El optimizador **Adam**.
   - b) La función de pérdida **Sparse Categorical Cross Entropy**.
   - c) La métrica **Accuracy**.

```python
model.compile(
    optimizer="adam",
    loss="sparse_categorical_crossentropy",
    metrics=["accuracy"]
)
```

8. Entrena el modelo, utilizando al menos 10 épocas.

```python
history = model.fit(
    X_train,
    y_train,
    epochs=10
)
```

### Evaluación

9. Evalúa el modelo utilizando el método correspondiente. ¿Crees que los resultados son satisfactorios?

```python
test_accuray, test_lost = model.evaluate(X_test, y_test)
```

10. Muestra **al menos diez imágenes** junto con la **etiqueta textual** de la categoría predicha por el modelo. Muestra, además, el **porcentaje de confianza** asignado por el mismo.

```python
import matplotlib.pyplot as plt
import numpy as np

# Mapeo de etiquetas numéricas a texto
labels_text = ['Camiseta', 'Pantalón', 'Pulóver', 'Vestido', 'Abrigo', 
               'Sandalia', 'Camisa', 'Zapatilla', 'Bolso', 'Botín']

# Obtener predicciones para el conjunto de prueba
predictions = model.predict(X_test)
predicted_classes = np.argmax(predictions, axis=1)
confidences = np.max(predictions, axis=1)

# Mostrar 10 imágenes del conjunto de prueba
fig, axes = plt.subplots(2, 5, figsize=(15, 7))
axes = axes.flatten()

for i in range(10):
    # Imagen, etiqueta real, predicción y confianza
    img = X_test[i]
    real_label = y_test[i]
    pred_label = predicted_classes[i]
    confidence = confidences[i] * 100  # Convertir a porcentaje
    
    # Mostrar imagen
    axes[i].imshow(img, cmap='gray')
    
    # Título con información
    color = 'green' if real_label == pred_label else 'red'
    axes[i].set_title(
        f'Real: {labels_text[real_label]}\n'
        f'Pred: {labels_text[pred_label]}\n'
        f'Conf: {confidence:.1f}%',
        fontsize=9,
        color=color
    )
    axes[i].axis('off')

plt.tight_layout()
plt.show()
```
